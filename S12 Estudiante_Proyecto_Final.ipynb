### 2.1 Cálculo de KPIs principales

**🎯 Objetivo:** Calcular los indicadores clave del negocio para evaluar ingresos, costos y rentabilidad.

```python
# Unir tablas 
orders_catalog = orders_clean.merge(catalog_clean[['nombre_producto','costo_unitario']], on='nombre_producto', how='left')
```

```python
# costo de producto de pedido
orders_catalog['costo_total_pedido'] = orders_catalog['cantidad'] * orders_catalog['costo_unitario']
```

```python
# KPIs de Rentabilidad
revenue = orders_catalog['monto_total'].sum()
costo_productos = orders_catalog['costo_total_pedido'].sum()
inversion_marketing = marketing_clean['gasto'].sum()
costo_total = costo_productos + inversion_marketing
profit = revenue - costo_total
margen_ganancia = (profit / revenue) * 100

print("=== PARTE 1: RENTABILIDAD DEL NEGOCIO ===")
print(f"Revenue (Ingresos Totales):   ${revenue:,.2f}")
print(f"Costo de Productos (COGS):   ${costo_productos:,.2f}")
print(f"Inversión en Marketing:      ${inversion_marketing:,.2f}")
print(f"Costo Total:                 ${costo_total:,.2f}")
print(f"Profit (Ganancia Neta):      ${profit:,.2f}")
print(f"Margen de Ganancia:          {margen_ganancia:.2f}%\n")
```

```python
# Comportamiento de Ventas
ticket_promedio = orders_catalog['monto_total'].mean()
cant_promedio = orders_catalog['cantidad'].mean()
ventas_por_prod = orders_catalog.groupby('nombre_producto')['cantidad'].sum()
prod_mas_vendido = ventas_por_prod.idxmax()
cant_mas_vendido = ventas_por_prod.max()

print("=== PARTE 2: COMPORTAMIENTO DE VENTAS ===")
print(f"Ticket Promedio (AOV):         ${ticket_promedio:,.2f}")
print(f"Cantidad Promedio por Orden:   {cant_promedio:.2f} unidades")
print(f"Producto más vendido:          {prod_mas_vendido} ({cant_mas_vendido:,.0f} unidades)")
```

```python
# PARTE 1: Totales del funnel
# ======================

query_totals = '''
SELECT 
    nombre_evento,
    COUNT(*) AS total_eventos,
    COUNT(DISTINCT id_usuario) AS usuarios_unicos,
    COUNT(DISTINCT id_sesion) AS sesiones_unicas
FROM events
GROUP BY nombre_evento
ORDER BY total_eventos DESC;
'''

totals = pd.read_sql(query_totals, con=engine)
totals
```

```python
# PARTE 2: Conversiones
# ======================

query_conversion = '''

WITH first_visits AS (
  SELECT DISTINCT id_usuario, pais
  FROM events
  WHERE nombre_evento = 'first_visit'
),
select_item AS (
  SELECT DISTINCT id_usuario, pais
  FROM events
  WHERE nombre_evento = 'select_item'
),
add_to_cart AS (
  SELECT DISTINCT id_usuario, pais
  FROM events
  WHERE nombre_evento = 'add_to_cart'
),
begin_checkout AS (
  SELECT DISTINCT id_usuario, pais
  FROM events
  WHERE nombre_evento = 'begin_checkout'
),
add_payment_info AS (
  SELECT DISTINCT id_usuario, pais
  FROM events
  WHERE nombre_evento = 'add_payment_info'
),
purchase AS (
  SELECT DISTINCT id_usuario, pais
  FROM events
  WHERE nombre_evento = 'purchase'
),
funnel_counts AS (
  SELECT
    fv.pais,  
    COUNT(fv.id_usuario) AS usuarios_first_visit,
    COUNT(si.id_usuario) AS usuarios_select_item,
    COUNT(a.id_usuario) AS usuarios_add_to_cart,
    COUNT(bc.id_usuario) AS usuarios_begin_checkout,
    COUNT(api.id_usuario) AS usuarios_add_payment_info,
    COUNT(p.id_usuario) AS usuarios_purchase
  FROM first_visits fv
  LEFT JOIN select_item si        
      ON fv.id_usuario = si.id_usuario AND fv.pais = si.pais
  LEFT JOIN add_to_cart a         
      ON fv.id_usuario = a.id_usuario AND fv.pais = a.pais
  LEFT JOIN begin_checkout bc     
      ON fv.id_usuario = bc.id_usuario AND fv.pais = bc.pais
  LEFT JOIN add_payment_info api  
      ON fv.id_usuario = api.id_usuario AND fv.pais = api.pais
  LEFT JOIN purchase p            
      ON fv.id_usuario = p.id_usuario AND fv.pais = p.pais
  GROUP BY fv.pais
)
SELECT
    pais,
    usuarios_first_visit,
    usuarios_select_item,
    usuarios_add_to_cart,
    usuarios_begin_checkout,
    usuarios_add_payment_info,
    usuarios_purchase,
    ROUND(usuarios_select_item * 100.0 / NULLIF(usuarios_first_visit, 0), 2) AS pct_select_item,
    ROUND(usuarios_add_to_cart * 100.0 / NULLIF(usuarios_first_visit, 0), 2) AS pct_add_to_cart,
    ROUND(usuarios_begin_checkout * 100.0 / NULLIF(usuarios_first_visit, 0), 2) AS pct_begin_checkout,
    ROUND(usuarios_add_payment_info * 100.0 / NULLIF(usuarios_first_visit, 0), 2) AS pct_add_payment_info,
    ROUND(usuarios_purchase * 100.0 / NULLIF(usuarios_first_visit, 0), 2) AS pct_purchase
FROM funnel_counts
ORDER BY pct_purchase DESC;
'''

conversion = pd.read_sql(query_conversion, con=engine)
conversion
```

## Resumen Tecnico

- El proceso presenta un flujo muy positivo y estable de retencion hasta que llegamos al paso de begin_checkout a add_payment_info y lo mismo ocurre con la transicion de add_payment_info a purchase, lo que sujiere que puede que exista algun tipo de dificultad o algo que no le gusta a laspersonas en add_payment_info y en purchase, hay que revisar en purchase que la totalidad del precio sea expuesto desde el principio y haya tarifas ocultas o extras que hagan que las persoans no quieran completar la compra o explorar cual es el motivo del bajon en la conversion y en add_payment_info revisar el diseño de UI/UX para serciorarse que el formulario sea claro y facil de diligenciar y explorar otros posibles inconvenientes o exigencias de la plataforma que puedan hacer perder clientes en este proceso.

## 🔹 Paso 4: Evaluar si los usuarios regresan (retención por cohortes)

**🎯 Objetivo:** Analizar la retención de usuarios para entender si regresan después de registrarse.

**Tablas**

- `users` 
- `user_activity` 

---
1. Se identifica la cohorte de cada usuario según el **mes de registro**.


2. Se calcula la retención semanal: cuántos usuarios **se mantienen activos** en cada semana desde su registro.
   - `retenido_w1`: usuarios activos en la semana 1  
   - `retenido_w2`: usuarios activos en la semana 2  
   - `retenido_w3`: usuarios activos en la semana 3  


3. Se calcula el porcentaje de retención para cada semana, dividiendo los usuarios retenidos entre los clientes iniciales de la cohorte:  
   - `semana_1`: porcentaje de usuarios retenidos en la semana 1  
   - `semana_2`: porcentaje de usuarios retenidos en la semana 2  
   - `semana_3`: porcentaje de usuarios retenidos en la semana 3  

Se revisa que la columna de fecha esté en formato correcto (`DATE`).  
Se realiza la conversión usando: `CAST(fecha_registro AS DATE)`

```python
# Explorar tabla users
# =========================
query_users = '''
SELECT *
FROM users;
'''
users = pd.read_sql(query_users, con=engine)
users.head(3)
```

```python
# Explorar tabla user_activity
# =========================
query_user_activity = '''
SELECT *
FROM user_activity;
'''
user_activity = pd.read_sql(query_user_activity, con=engine)
user_activity.head(3)
```

```python
# Retención por cohortes
# ======================

query_cohort_retention_final = '''

WITH cohort AS (
    SELECT
        id_usuario,
        TO_CHAR(DATE_TRUNC('month', MIN(CAST(fecha_registro AS DATE))), 'YYYY-MM') AS cohort
    FROM users
    GROUP BY id_usuario
),
activity AS (
    SELECT 
        u.id_usuario, 
        c.cohort, 
        ua.dias_despues_registro,
        ua.activo
    FROM users u
    LEFT JOIN cohort c 
        ON u.id_usuario = c.id_usuario
    LEFT JOIN user_activity ua 
        ON u.id_usuario = ua.id_usuario
)
SELECT 
    cohort,
    COUNT(DISTINCT id_usuario) AS usuarios_cohorte,
    
    COUNT(DISTINCT CASE WHEN dias_despues_registro >= 7 AND activo = 1 THEN id_usuario END) AS retenidos_d7,
    COUNT(DISTINCT CASE WHEN dias_despues_registro >= 14 AND activo = 1 THEN id_usuario END) AS retenidos_d14,
    COUNT(DISTINCT CASE WHEN dias_despues_registro >= 21 AND activo = 1 THEN id_usuario END) AS retenidos_d21,
    COUNT(DISTINCT CASE WHEN dias_despues_registro >= 28 AND activo = 1 THEN id_usuario END) AS retenidos_d28,
    
    ROUND(
      COUNT(DISTINCT CASE WHEN dias_despues_registro >= 7 AND activo = 1 THEN id_usuario END) * 100.0 
      / NULLIF(COUNT(DISTINCT id_usuario), 0), 1
    ) AS retention_d7_pct,
    
    ROUND(
      COUNT(DISTINCT CASE WHEN dias_despues_registro >= 14 AND activo = 1 THEN id_usuario END) * 100.0 
      / NULLIF(COUNT(DISTINCT id_usuario), 0), 1
    ) AS retention_d14_pct,
    
    ROUND(
      COUNT(DISTINCT CASE WHEN dias_despues_registro >= 21 AND activo = 1 THEN id_usuario END) * 100.0 
      / NULLIF(COUNT(DISTINCT id_usuario), 0), 1
    ) AS retention_d21_pct,
    
    ROUND(
      COUNT(DISTINCT CASE WHEN dias_despues_registro >= 28 AND activo = 1 THEN id_usuario END) * 100.0 
      / NULLIF(COUNT(DISTINCT id_usuario), 0), 1
    ) AS retention_d28_pct

FROM activity
GROUP BY cohort
ORDER BY cohort;
'''

# Ejecutar la consulta
cohorte_final = pd.read_sql(query_cohort_retention_final, con=engine)
cohorte_final
```

Muy buena sección de cohortes.

Partes de `fecha_registro` para construir la cohorte mensual y conectas correctamente cada usuario con `user_activity`. Además, la lógica de retención utiliza condiciones acumuladas como <code>&gt;= 7</code>, <code>&gt;= 14</code>, <code>&gt;= 21</code> y <code>&gt;= 28</code>, manteniendo como denominador el tamaño inicial de cada cohorte.

La salida también es fácil de interpretar y muestra un patrón consistente: retención alta alrededor de los primeros 7 días y una disminución progresiva hacia horizontes posteriores.

## 🔹 Paso 5: Validar si los cambios generan impacto (test estadístico)

🎯 **Objetivo:** Evaluar si la modificación en la UI del checkout impacta la **tasa de conversión de compra**.

---

1. **Analizar el dataset** `experiment_checkout_ui.csv` para identificar la métrica principal **conversion**.
   - La métrica **conversion** es 1 si el usuario completó la compra, 0 si no.    
2. **Plantear la hipótesis estadística**     
3. **Aplicar el test estadístico adecuado** 
4. **Interpretar el resultado**  

---
Hipótesis estadística
   - **H₀ (Hipótesis nula):** No existe una diferencia significativa en la tasa de conversión entre el grupo de control y el grupo de tratamiento
   - **H₁ (Hipótesis alternativa):** Existe una diferencia significativa en la tasa de conversión entre el grupo de control y el grupo de tratamiento
   
**Test estadístico:** Prueba Z o pruba de chi_2.
**Nivel de significancia alpha:** 0.05 o (5%)

```python
# tu código aquí
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from scipy.stats import chi2_contingency, levene, ttest_ind

pd.set_option("display.max_columns", 200)
pd.set_option("display.width", 140)
```

```python
data = pd.read_csv('https://practicum-content.s3.amazonaws.com/datasets/experiment_checkout_ui.csv')
print("Dimensiones del dataset:", data.shape)
data.head()
```

```python
data.info()
```

```python
# Nulos por columna
nulls = data.isna().sum()
display(nulls)

# Duplicados por usuario
duplicated_users = data.duplicated(subset=["id_usuario"]).sum()
print("Usuarios duplicados:", duplicated_users)
```

```python
categorical_cols = ['id_usuario', 'variante', 'convirtio', 'dispositivo', 'pais']

for col in categorical_cols:
    print("\n---", col, "---")
    display(data[col].value_counts())
```

```python
clean = data.copy()

# Normalizar categorías de texto
clean["convirtio"] = clean["convirtio"].astype(int)
clean["variante"] = clean["variante"].astype(str).str.strip().str.title()
clean["dispositivo"] = clean["dispositivo"].astype(str).str.strip().str.title()

# Convertir fecha
clean["timestamp"] = pd.to_datetime(clean["timestamp"], errors="coerce")

# Corregir valores imposibles
clean = clean[clean["duracion_sesion"] > 0]

print("Dimensiones antes:", data.shape)
print("Dimensiones después:", clean.shape)

display(clean.head())
```

```python
print("Distribución:")
display(clean["dispositivo"].value_counts())
display(clean["variante"].value_counts())
display(clean["convirtio"].value_counts())

print("\nNulos restantes:")
display(clean.isna().sum())

print("\nRango de fechas:")
print(clean["timestamp"].min(), "→", clean["timestamp"].max())
```

```python
conversion_by_landing = clean.groupby("variante")["convirtio"].agg(
    users="count",
    conversions="sum", 
    conversion_rate="mean"
).reset_index()

display(conversion_by_landing)
```

```python
# Gráfica simple con .plot()


df_conversion = conversion_by_landing.copy()
df_conversion['conversion_pct'] = df_conversion['conversion_rate'] * 100

# Estilo visual
sns.set_theme(style="whitegrid")
plt.figure(figsize=(7, 5))

# Graficar barras
bars = plt.bar(
    df_conversion['variante'], 
    df_conversion['conversion_pct'], 
    color=['#34495e', '#2ecc71'], 
    width=0.45, 
    edgecolor='black', 
    linewidth=0.8
)

# Títulos y etiquetas
plt.title('Tasa de Conversión por Variante (A/B Test)', fontsize=14, fontweight='bold', pad=15)
plt.ylabel('Tasa de Conversión (%)', fontsize=12)
plt.xlabel('Variante UI', fontsize=12)
plt.ylim(0, 22)

# Añadir etiquetas de porcentaje sobre cada barra
for bar in bars:
    yval = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width()/2.0, 
        yval + 0.6, 
        f'{yval:.2f}%', 
        ha='center', 
        va='bottom', 
        fontsize=11, 
        fontweight='bold'
    )

plt.axhline(y=df_conversion['conversion_pct'].mean(), color='gray', linestyle='--', alpha=0.5, label='Promedio')
plt.legend(loc='upper right')
plt.tight_layout()
plt.show()
```

```python

# Crear la tabla de contingencia entre variante y convirtió
contingency_table = pd.crosstab(clean["variante"], clean["convirtio"])
display(contingency_table)

# test de Chi-cuadrado
chi2, p_value, dof, expected = chi2_contingency(contingency_table)

print(f"Estadístico Chi2: {chi2:.4f}")
print(f"p-value: {p_value:.5f}")

# Conclusion
alpha = 0.05
if p_value < alpha:
    print("Rechazamos H0: Hay un impacto estadísticamente significativo en la tasa de conversión.")
else:
    print("No rechazamos H0: No hay evidencia suficiente de que el cambio en la UI impacte la conversión.")
```
